### Parsers

## Encadeamento

```python
chain = prompt_template | llm | StrOutputParser()
```

```python
response = chain.invoke({
    "lingua_origem": "inglês",
    "lingua_destino": "espanhol",
    "texto": "As praias de Recife tem tubarões!"
})

print(response)
```

```python
def translate(texto, lingua_origem, lingua_destino):
    response = chain.invoke({
        "lingua_origem": lingua_origem,
        "lingua_destino": lingua_destino,
        "texto": texto
    })
    return response
```

```python
output = translate("The beaches of Recife have sharks!", "inglês", "espanhol")

print(output)
```

### Exercício 1
Crie uma `chain` que a partir de um tópico informado pelo usuário, crie uma piada.

```python
# Template para gerar piadas
prompt_template = """
Você é um gênio da piada e precisa criar uma piada relevante para o seguinte tópico:
{tópico}

Sua resposta deve ser uma única frase curta que seja uma piada relacionada ao tópico fornecido. Por exemplo:
"Por que os programadores usam variáveis? Porque elas são mutáveis!"

Se você não conseguir pensar em uma piada relevante, diga "Não tenho uma piada sobre isso".

Tópico: {tópico}
"""

# Função para gerar piadas usando LangChain
def get_piada_langchain(tópico):
    try:
        prompt = prompt_template.format(tópico=tópico)
        messages = [
            {"role": "system", "content": "Você é um gênio da piada."},
            {"role": "user", "content": prompt}
        ]

        # Chama o modelo usando invoke
        response = llm.invoke(input=messages)

        # Retorna o conteúdo gerado
        return response.content.strip()
    except Exception as e:
        print("Erro dentro da função get_piada_langchain:", e)
        raise

# Exemplo de uso
while True:
    t = input("Digite um tópico para uma piada: ")
    try:
        piada = get_piada_langchain(t)
        print("Piada gerada:", piada)
    except Exception as e:
        print(f"Erro ao gerar piada: {e}")

    resposta = input("Deseja ouvir outra piada? (S/N): ").strip().upper()
    if resposta != 'S':
        print("Saindo...")
        break
```

### Exercício 2
Crie uma `chain` que classifique o sentimento de um texto de entrada em positivo, neutro ou negativo.

```python
from langchain.prompts import PromptTemplate
# Prompt para classificação de sentimento
sentimento_prompt = PromptTemplate(
    input_variables=["texto"],
    template="""
    Você é um especialista em análise de sentimentos. Analise o seguinte texto e classifique o sentimento como positivo, neutro ou negativo.

    Texto: "{texto}"
    Classificação do sentimento:
    """
)

# Função para classificar o sentimento usando LangChain
def classificar_sentimento(texto):
    try:
        prompt = sentimento_prompt.format(texto=texto)
        messages = [
            {"role": "system", "content": "Você é um especialista em análise de sentimentos."},
            {"role": "user", "content": prompt}
        ]


        # Chama o modelo
        response = llm.invoke(input=messages)


        # Retorna o conteúdo gerado
        return response.content.strip()
    except Exception as e:
        print("Erro dentro da função classificar_sentimento:", e)
        raise

# Exemplo de uso
if __name__ == "__main__":
    while True:
        texto = input("Digite um texto para análise de sentimento: ")
        try:
            classificacao = classificar_sentimento(texto)
            print("Classificação do sentimento:", classificacao)
        except Exception as e:
            print(f"Erro ao classificar sentimento: {e}")

        resposta = input("Deseja analisar outro texto? (S/N): ").strip().upper()
        if resposta != 'S':
            print("Saindo...")
            break
```

### Exercício 3
Crie uma `chain` que gere o código de uma função Python de acordo com a descrição do usuário.

```python
codigo_prompt = PromptTemplate(
    input_variables=["descricao"],
    template="""
    Você é um especialista em programação Python. Com base na seguinte descrição, gere o código de uma função Python que atenda aos requisitos:

    Descrição: "{descricao}"

    Código da função:
    """
)

# Função para gerar o código Python a partir da descrição
def gerar_codigo_funcao(descricao):
    try:
        prompt = codigo_prompt.format(descricao=descricao)
        messages = [
            {"role": "system", "content": "Você é um especialista em programação Python."},
            {"role": "user", "content": prompt}
        ]


        # Chama o modelo
        response = llm.invoke(input=messages)


        # Retorna o código gerado
        return response.content.strip()
    except Exception as e:
        print("Erro dentro da função gerar_codigo_funcao:", e)
        raise

if __name__ == "__main__":
    while True:
        descricao = input("Descreva a função Python que você deseja gerar: ")
        try:
            codigo = gerar_codigo_funcao(descricao)
            print("Código gerado:\n", codigo)
        except Exception as e:
            print(f"Erro ao gerar código: {e}")

        resposta = input("Deseja gerar outra função? (S/N): ").strip().upper()
        if resposta != 'S':
            print("Saindo...")
            break
```

### Exercício 4
Crie uma `chain` que explique de forma simplificada um tópico geral fornecido pelo usuário e, em seguida, traduza a explicação para inglês. Utilize dois templates encadeados.

```python
# Template para explicar o tópico
explicacao_prompt = PromptTemplate(
    input_variables=["topico"],
    template="""
    Você é um especialista em simplificar conceitos complexos. Explique de forma simples o seguinte tópico:

    Tópico: {topico}

    Explicação simplificada:
    """
)

# Template para traduzir para inglês
traducao_prompt = PromptTemplate(
    input_variables=["explicacao"],
    template="""
    Traduza o seguinte texto para o inglês:

    Texto: {explicacao}

    Tradução:
    """
)

# Função para explicar e traduzir um tópico
def explica_e_traduz(topico):
    try:
        # Passo 1: Explicação simplificada
        explicacao_messages = [
            {"role": "system", "content": "Você é um especialista em simplificar conceitos complexos."},
            {"role": "user", "content": explicacao_prompt.format(topico=topico)}
        ]

        explicacao_response = llm.invoke(input=explicacao_messages)
        explicacao_simplificada = explicacao_response.content.strip()


        # Passo 2: Tradução para inglês
        traducao_messages = [
            {"role": "system", "content": "Você é um tradutor profissional."},
            {"role": "user", "content": traducao_prompt.format(explicacao=explicacao_simplificada)}
        ]

        traducao_response = llm.invoke(input=traducao_messages)
        traducao_final = traducao_response.content.strip()


        return explicacao_simplificada, traducao_final
    except Exception as e:
        print("Erro na função explica_e_traduz:", e)
        raise

# Exemplo de uso
if __name__ == "__main__":
    while True:
        topico = input("Digite um tópico para explicar e traduzir: ")
        try:
            explicacao, traducao = explica_e_traduz(topico)
            print("\nExplicação simplificada:", explicacao)
            print("\nTradução para inglês:", traducao)
        except Exception as e:
            print(f"Erro ao processar o tópico: {e}")

        resposta = input("Deseja explicar e traduzir outro tópico? (S/N): ").strip().upper()
        if resposta != 'S':
            print("Saindo...")
            break
```

### Exercício 5 - Desafio
Crie uma `chain` que responda perguntas sobre o CESAR School.

```python
# Prompt para responder perguntas sobre o CESAR School
cesar_prompt = PromptTemplate(
    input_variables=["pergunta"],
    template="""
    Você é um especialista no CESAR School. Responda de forma clara e objetiva à seguinte pergunta:

    Pergunta: {pergunta}

    Resposta:
    """
)

# Função para responder perguntas sobre o CESAR School
def responder_pergunta_cesar(pergunta):
    try:
        prompt = cesar_prompt.format(pergunta=pergunta)
        messages = [
            {"role": "system", "content": "Você é um especialista no CESAR School e conhece todos os detalhes sobre a instituição."},
            {"role": "user", "content": prompt}
        ]

        # Chama o modelo
        response = llm.invoke(input=messages)

        # Retorna a resposta gerada
        return response.content.strip()
    except Exception as e:
        print("Erro dentro da função responder_pergunta_cesar:", e)
        raise

# Exemplo de uso
if __name__ == "__main__":
    while True:
        pergunta = input("Digite sua pergunta sobre o CESAR School: ")
        try:
            resposta = responder_pergunta_cesar(pergunta)
            print("\nResposta:", resposta)
        except Exception as e:
            print(f"Erro ao responder a pergunta: {e}")

        resposta_continuar = input("Deseja fazer outra pergunta? (S/N): ").strip().upper()
        if resposta_continuar != 'S':
            print("Saindo...")
            break
```
